# 02 Source-label spectral connectivity

Compute spectral connectivity from epoch-wise source-label time courses. Start with one recording as a smoke test.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent
elif PROJECT_ROOT.name == "4_connectivity":
    PROJECT_ROOT = PROJECT_ROOT.parents[1]

CONFIG_PATH = PROJECT_ROOT / "configs" / "local.yaml"
print("PROJECT_ROOT:", PROJECT_ROOT)
print("CONFIG_PATH:", CONFIG_PATH)

from meeg_pipeline.config import load_config
from meeg_pipeline.workflow import iter_recordings, recordings_to_dataframe

config = load_config(CONFIG_PATH)
recordings = list(iter_recordings(config, subjects="all"))
recordings_to_dataframe(recordings)
```

```python
from meeg_pipeline.connectivity import (
    connectivity_input_overview_to_dataframe,
    compute_source_label_connectivity_for_recordings,
    connectivity_results_to_dataframe,
)

ON_EXISTING = "skip"  # "skip" | "overwrite" | "error"
MAX_RECORDINGS_FOR_TEST = None  # set to None for all recordings
METHODS = None               # None -> config.connectivity.methods
WINDOWS = None               # None -> config.connectivity.windows
CONDITIONS = None            # None -> config.connectivity.conditions
VERBOSE = False

selected_recordings = recordings[:MAX_RECORDINGS_FOR_TEST] if MAX_RECORDINGS_FOR_TEST else recordings
connectivity_input_overview_to_dataframe(config, selected_recordings, on_existing=ON_EXISTING)
```

```python
results = compute_source_label_connectivity_for_recordings(
    config,
    selected_recordings,
    on_existing=ON_EXISTING,
    methods=METHODS,
    windows=WINDOWS,
    conditions=CONDITIONS,
    verbose=VERBOSE,
)

results_df = connectivity_results_to_dataframe(results)
results_df
```
